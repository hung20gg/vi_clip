```python
from llm import Gemini


llm = Gemini(model_name="gemini-2.0-flash-lite")


messages = [
    {
        'role': 'system',
        'content' : 'Use 10 word only'
    },
    {
        'role': 'user',
        'content' : 'Who is Elon Musk?'
    }
]
```

```python
messages_lst = [messages for _ in range(10)]
```

```python
llm.batch_call(messages_lst, prefix = 'test', example_per_batch=5)
```

```python
batch_ids = llm.recall_local_batch('test')
```

```python
batch_ids
```

```python
results = llm.get_successful_messages(batch_ids)
```
